Implement built method Sum in python

```python
def custom_sum(*args, starting=0):
    output = starting
    for item in args:
        output += item
        
    return output


custom_sum(1,2,3)
```

```python
def custom_avg(*args, starting=0):
    output = starting
    for item in args:
        output += item
        
    return output//len(args)


custom_avg(1,2,3)
```

Write a function that takes a list of words (strings). It should return a tuple containing three integers, representing the length of the shortest word, the length
of the longest word, and the average word length.

```python
def list_of_words(words_list):
    shortest = len(words_list[0])
    longest = len(words_list[1])
    mysum = 0
    for item in words_list:
        if len(item) > shortest:
            longest = len(item)
        else:
            shortest = len(item)
            
        mysum += len(item)
    return (shortest,longest,mysum//len(words_list))

list_of_words(['aaa', 'bb', 'cccc'])
```
