Write a function (guessing_game) that takes no arguments.
When run, the function chooses a random integer b
(inclusive).
Then ask the user to guess what number has been chosen.
Each time the user enters a guess, the program indicates one of the following:
– Too high
– Too low
– Just right
If the user guesses correctly, the program exits. Otherwise, the user is asked to
try again.
The program only exits after the user guesses correctly.

```python
import random
def gussing_game():
    random_num = random.randint(1,10)
    program_cont = True
    
    while program_cont:
        user_guess = int(input("Please Guess the Number"))
        if user_guess > random_num:
            print("Number you guessed is Too High")
        elif user_guess < random_num:
            print("Guess is too low")
        elif user_guess == random_num:
            program_cont = False
            print("Tadaw, You have guessed Right")

gussing_game()
```
