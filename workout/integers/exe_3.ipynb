Running Time

```python
def run_time():
    average_time = 0
    total_run = 0
    while True:
        
        run_mints = input('Enter 10KM run time: ')
        if not run_mints:
            break
        total_run += 1
        average_time += float(run_mints)
        
    return f'Average of {average_time/total_run:.2f}, over {total_run}'
print(run_time())
```

Write a function that takes a float and two integers (before and after). The
function should return a float consisting of before digits before the decimal
point and after digits after. Thus, if we call the function with 1234.5678, 2, and
3, the return value should be 34.567.

```python
def slice_float(floating,before,after):
    int_part, float_part = str(floating).split('.',1)
    new_int = int_part[-before:] if len(int_part) >= before else int_part
    new_float = float_part[:after] if len(float_part) >= after else float_part
    
    return f'{new_int}.{new_float}' if new_float else new_int


print(slice_float(1234.5678, 2, 3))  # 34.567
print(slice_float(1234.5678, 1, 1))  # 4.5
print(slice_float(1234.5678, 4, 2))  # 1234.56
print(slice_float(-9876.54321, 2, 2)) # -76.54
print(slice_float(12.3, 5, 5))
```

decimal number can be. Write a function that takes two strings from the user,
turns them into decimal instances, and then prints the floating-point sum of the
user’s two inputs. In other words, make it possible for the user to enter 0.1 and
0.2, and for us to get 0.3 back.

```python
from decimal import Decimal
def decimal_number():
    s1 = input('Enter First Number').strip()
    s2 = input('Enter Second Number').strip()
    
    result = Decimal(s1) + Decimal(s2)
    
    return float(result)

print(f"Exact Decimal result = {decimal_number()}")
```
