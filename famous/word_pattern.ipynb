Given a pattern and a string s, return true if the string s follows the same pattern.

Here, the following the pattern means each character in the pattern should map to a single word in s, and each word in s should map to a single character in the pattern.
Input: pattern = "eegg", s = "dog dog cat cat"
Output: true
Explanation: The pattern "eegg" corresponds to the words "dog dog cat cat". Both 'e's map to "dog" and both 'g's map to "cat".

```python
def word_pattern(pattern,s):
    
    char_word = {}
    word_char = {}
    s_list = s.split()
    
    if len(s_list) != len(pattern): return False
    
    for ch, word in zip(pattern,s_list):
        
        if ch in char_word:
            if char_word[ch] != word:
                return False
        else:
            char_word[ch] = word
            
        if word in word_char:
            if word_char[word] != ch:
                return False
        else:
            word_char[word] = ch

    return True
    
    
word_pattern(pattern = "abacac", s = "dog cat dog mouse dog mouse")
word_pattern(pattern = "abca", s = "one two three four")
word_pattern(pattern = "eegg", s = "dog dog cat cat")
```

Given an unsorted array of integers, find the length of the longest consecutive sequence of numbers in it. A consecutive sequence means the numbers in the sequence are contiguous without any gaps. For instance, 1, 2, 3, 4 is a consecutive sequence, but 1, 3, 4, 5 is not.

```python
def longest_seq(input):
    sorted_list = set(input)
    longest = 0
    for num in sorted_list:
        if num - 1 not in sorted_list:
            current = num
            length = 1
            
            while current+1 in sorted_list:
                current +=1
                length +=1
            longest = max(length,longest)
            
    return longest
```
